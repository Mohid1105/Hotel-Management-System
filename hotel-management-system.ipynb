```python
room_numbers = [101,102,103,201,202,203,301,302,303]
room_types = ["single", "suite", "double", "single", "single", "double", "suite", "suite", "single"]
room_prices = [100,300,200,100,100,200,300,300,100]
room_availablility = ["available", "available", "available", "available", "available", "available", "available", "available", "available"]

floor_1 = [101,102,103]
floor_2 = [201,202,203]
floor_3 = [301,302,303]

guest_names = []
guest_ages = []
guest_catergories = []

booking_guest_name = []
booking_nights = []
booking_room_numb = []
booking_services = []

services_menu = ["spa","breakfast","lunch","dinner","pool","massage"]
services_prices = [50,30,45,45,35,50]

total_revenue = 0

while True:

    option = input("1:view all rooms\n2:register guest\n3:make reservations\n4:checking guest\n5:order food and services\n6:checkout and generate bill\n7:cancel reservations\n8:hotel statistics\n9:exit\n")

    # VIEW ALL ROOMS
    if option == "1":

        for i in range(len(room_numbers)):
            print(f"Room number:{room_numbers[i]} | Room type:{room_types[i]} | Room price:{room_prices[i]} | Room availability:{room_availablility[i]}")
            print()


    # REGISTER GUEST
    elif option == "2":

        print("="*20 + " Guest registration " + "="*20)

        name = input("Enter guest name: ")
        age = int(input("Enter guest age: "))

        if age < 18:
            print("You must be 18 or older to register as guest")

        else:

            cate_choice = input("1:regular\n2:VIP\n3:corporate\nEnter choice: ")

            if cate_choice == "1":
                category = "regular"

            elif cate_choice == "2":
                category = "VIP"

            elif cate_choice == "3":
                category = "corporate"

            else:
                print("Invalid category")
                continue

            if category == "VIP":

                guest_names.insert(0, name)
                guest_ages.insert(0, age)
                guest_catergories.insert(0, category)

            else:

                guest_names.append(name)
                guest_ages.append(age)
                guest_catergories.append(category)

            print("Guest has been registered")


    # MAKE RESERVATION
    elif option == "3":

        print("="*20 + " Making reservation " + "="*20)

        name = input("Enter guest name: ")

        guest_found = False

        for i in range(len(guest_names)):

            if guest_names[i] == name:
                guest_found = True
                break

        if not guest_found:

            print("Guest name is not registered. Please register first")

        else:

            nights = int(input("Enter number of nights: "))

            if nights <= 0:

                print("Invalid. Enter correct number of nights")

            else:

                room_choice = input("1:single\n2:double\n3:suite\nEnter choice: ")

                if room_choice == "1":
                    room_type = "single"

                elif room_choice == "2":
                    room_type = "double"

                elif room_choice == "3":
                    room_type = "suite"

                else:
                    room_type = ""

                if room_type == "":

                    print("Invalid choice")

                else:

                    print("Available " + room_type + " rooms:")

                    for i in range(len(room_numbers)):

                        if room_types[i] == room_type and room_availablility[i] == "available":

                            print(f"Room {room_numbers[i]} | RS {room_prices[i]}")

                    room_num = int(input("Enter room number to reserve: "))

                    room_index = -1

                    for i in range(len(room_numbers)):

                        if room_numbers[i] == room_num:
                            room_index = i
                            break

                    if room_index == -1:

                        print("Invalid room number")

                    elif room_types[room_index] != room_type:

                        print("This room is not the room type you want")

                    elif room_availablility[room_index] != "available":

                        print("This room is not available. Pick again")

                    else:

                        booking_guest_name.append(name)
                        booking_nights.append(nights)
                        booking_room_numb.append(room_num)
                        booking_services.append([])

                        room_availablility[room_index] = "reserved"

                        subtotal = room_prices[room_index] * nights

                        print(f"Reservation for {name} room:{room_num} for number of nights:{nights}")
                        print(f"The total is {subtotal}")


    # CHECK IN
    elif option == "4":

        print("="*20 + " Check in " + "="*20)

        name = input("Enter your name: ")
        room_num = int(input("Enter the booked room number: "))

        booking_index = -1

        for i in range(len(booking_guest_name)):

            if booking_guest_name[i] == name and booking_room_numb[i] == room_num:
                booking_index = i
                break

        if booking_index == -1:

            print("No reservations found")

        else:

            room_index = -1

            for i in range(len(room_numbers)):

                if room_numbers[i] == room_num:
                    room_index = i
                    break

            if room_availablility[room_index] == "occupied":

                print("Room is already occupied")

            else:

                room_availablility[room_index] = "occupied"

                print("Check in successful")
                print(f"Guest:{name}\nRoom:{room_num}")


    # FOOD AND SERVICES
    elif option == "5":

        print("="*20 + " Hotel services " + "="*20)

        room_num = int(input("Enter your room number: "))

        booking_index = -1

        for i in range(len(booking_room_numb)):

            if booking_room_numb[i] == room_num:
                booking_index = i
                break

        if booking_index == -1:

            print("No active booking for this room")

        else:

            while True:

                print("\nServices")

                for i in range(len(services_menu)):
                    print(f"{i+1}. {services_menu[i]} | ${services_prices[i]}")

                print("7. finish")

                service_choice = int(input("Enter the desired service: "))

                if service_choice == 7:

                    break

                elif service_choice >= 1 and service_choice <= len(services_menu):

                    chosen_service = services_menu[service_choice - 1]

                    booking_services[booking_index].append(chosen_service)

                    print(f"{chosen_service} has been added")

                else:

                    print("Invalid choice")
                    continue

                total_services = 0

                for service in booking_services[booking_index]:

                    for i in range(len(services_menu)):

                        if services_menu[i] == service:

                            total_services += services_prices[i]
                            break

                print(f"Services total bill is ${total_services}")


    # CHECKOUT
    elif option == "6":

        print("="*20 + " Checkout " + "="*20)

        room_num = int(input("Enter your room number: "))

        booking_index = -1

        for i in range(len(booking_room_numb)):

            if booking_room_numb[i] == room_num:
                booking_index = i
                break

        if booking_index == -1:

            print("Invalid booking number")

        else:

            room_index = -1

            for i in range(len(room_numbers)):

                if room_numbers[i] == room_num:
                    room_index = i
                    break

            guest_name = booking_guest_name[booking_index]
            nights = booking_nights[booking_index]

            room_charges = room_prices[room_index] * nights

            services_total = 0

            for service in booking_services[booking_index]:

                for i in range(len(services_menu)):

                    if services_menu[i] == service:

                        services_total += services_prices[i]
                        break

            subtotal = room_charges + services_total

            tax = subtotal * 0.05

            total = subtotal + tax

            print("="*75)
            print("                   FINAL BILL")
            print("="*75)

            print(f"Guest name: {guest_name}")
            print(f"Room: {room_num}")
            print(f"Nights: {nights}")
            print()

            print(f"Room charges: ${room_charges}")
            print(f"Food and services: ${services_total}")
            print(f"Subtotal: ${subtotal}")
            print(f"Tax (5%): ${tax}")
            print(f"Total: ${total}")

            print()

            print("Payment method")
            payment_method = input("1:cash\n2:card\nEnter here: ")

            payment = float(input("Enter the payment amount: $"))

            if payment < total:

                remaining = total - payment

                print(f"The amount is insufficient. Remaining amount: ${remaining}")

            else:

                if payment > total:

                    change = payment - total

                    print(f"Payment successful. Your change is: ${change}")

                else:

                    print("Payment successful. Exact amount received")

                total_revenue += total

                room_availablility[room_index] = "available"

                booking_guest_name.pop(booking_index)
                booking_room_numb.pop(booking_index)
                booking_nights.pop(booking_index)
                booking_services.pop(booking_index)


    # CANCEL RESERVATION
    elif option == "7":

        print("="*20 + " Cancel reservation " + "="*20)

        room_num = int(input("Enter room number: "))

        booking_index = -1

        for i in range(len(booking_room_numb)):

            if booking_room_numb[i] == room_num:
                booking_index = i
                break

        if booking_index == -1:

            print("No reservations found")

        else:

            print("Reservation is found")

            confirm = input("Are you sure you want to cancel the reservation?\n1:yes\n2:no\n")

            if confirm == "1":

                room_index = -1

                for i in range(len(room_numbers)):

                    if room_numbers[i] == room_num:
                        room_index = i
                        break

                room_availablility[room_index] = "available"

                booking_guest_name.pop(booking_index)
                booking_room_numb.pop(booking_index)
                booking_services.pop(booking_index)
                booking_nights.pop(booking_index)

                print("Your reservation has been cancelled")

            else:

                print("Cancel aborted")


    # HOTEL STATISTICS
    elif option == "8":

        available_count = 0
        occupied_count = 0

        for i in range(len(room_availablility)):

            if room_availablility[i] == "available":

                available_count += 1

            else:

                occupied_count += 1

        vip_count = 0

        for i in range(len(guest_catergories)):

            if guest_catergories[i] == "VIP":

                vip_count += 1

        cheapest_price = room_prices[0]
        cheapest_type = room_types[0]

        most_expensive = room_prices[0]
        most_expensive_type = room_types[0]

        for i in range(len(room_prices)):

            if room_prices[i] < cheapest_price:

                cheapest_price = room_prices[i]
                cheapest_type = room_types[i]

            if room_prices[i] > most_expensive:

                most_expensive = room_prices[i]
                most_expensive_type = room_types[i]

        print("="*30 + " Hotel statistics " + "="*30)

        print(f"Total rooms: {len(room_numbers)}")
        print(f"Available rooms: {available_count}")
        print(f"Occupied rooms: {occupied_count}")
        print(f"Total guest: {len(guest_names)}")
        print(f"VIP guests: {vip_count}")
        print(f"Total revenue: {total_revenue}")
        print(f"Most expensive room type: {most_expensive_type}")
        print(f"Cheapest room type: {cheapest_type}")


    # EXIT
    elif option == "9":

        print("Thank you for using Mohid Grand Hotel management system.")

        break


    else:

        print("Choose a valid option")
```
